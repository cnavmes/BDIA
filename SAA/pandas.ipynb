# **.describe()**
---

Si la columna es numérica:
Te da estadísticas básicas como:
- count: número de valores no nulos
- mean: media
- std: desviación estándar
- min: mínimo
- 25%: percentil 25
- 50%: percentil 50 (la mediana)
- 75%: percentil 75
- max: máximo

Si la columna es de tipo objeto o categórica:
- count: número de valores no nulos
- unique: número de valores únicos
- top: valor más frecuente
- freq: frecuencia del valor más frecuente

```python
import pandas as pd

stocks = pd.read_csv("stocks.csv")
stocks.head(2)
```

```python
stocks.describe()
```

```python
perc = [.50, .90]
subsets = pd.IndexSlice[['90%'], 'Open']
stocks.describe(percentiles = perc).style.map(lambda x: "background-color: magenta", subset=subsets)
	# 1.	Calcula estadísticas de stocks con percentiles 50% y 90%.
	# 2.	Selecciona la celda del percentil 90 de la columna Open.
	# 3.	Pinta esa celda de magenta en la salida de Jupyter Notebook.
```

# **.apply()**
---
Sirve para aplicar una función a lo largo de un eje de un DataFrame o a cada elemento de una Series.

- axis=0 (por defecto): aplica la función columna por columna.
- axis=1: aplica la función fila por fila.

```python
import pandas as pd
import numpy as np
df = pd.DataFrame({'A': [1,2], 'B':[4,5]})
df
```

```python
df.apply(func = np.sum, axis = 0)
```

```python
df.apply(func = np.sum, axis=1)
```

```python
A_plus_B = df.apply(func = lambda row: row['A'] + row['B'], axis=1)
df['A_plus_B'] = A_plus_B
df
```

```python
df.apply(func = lambda col: (col[0] + col[1])/2, axis=0)
```

# .groupby()
---

Sirve para agrupar filas de un DataFrame en función de una o varias columnas, y luego aplicar cálculos sobre cada grupo.

```python
import pandas as pd

df = pd.read_csv('weather.csv')
df.head(2)
```

```python
for city, city_df in df.groupby('Station.City'):
    print(city)
    print(city_df)
```

```python
df.groupby('Station.City').max()
```

```python
df.groupby('Station.City').describe()
```

# .iterrows() | .itertuples()
---

- .iterrows(): 
    - Devuelve cada fila como una tupla (índice, fila)
    - Donde fila es una pandas Series
- .itertuples():
    - Devuelve cada fila como un namedtuple (objeto tipo tupla con nombres de campo)
    - Es más rápido y eficiente que iterrows()

```python
import pandas as pd
df_penguins = pd.read_csv('penguins.csv')
```

```python
%%time
index_series_pairs = [pair for pair in df_penguins.iterrows()]
idx, series = index_series_pairs[0]
print('idx', idx)
print('__________________________')
print(series)
print('________time taken________')
```

```python
%%time
named_tuples = [tup for tup in df_penguins.itertuples()]
print(named_tuples[0])
print('________time taken________')
```

# .loc[]
---
Se usa para acceder a filas y columnas por sus etiquetas (nombres), no por posición numérica.


```python
import pandas as pd
df = pd.read_csv('penguins.csv')
df.head(4)
```

```python
#Usando un índice
df.loc[2]
```

```python
#slice
df.loc[0:3:2]
```

```python
df.loc[1:2, ['species', 'sex']]
```

```python
df.loc[(df['bill_length_mm'] > 50.5) & (df['sex'] == 'female')]
```

# .sort_values()
---

Sirve para ordenar las filas de un DataFrame o los elementos de una Serie según una o varias columnas

```python
import pandas as pd
df = pd.read_csv('penguins.csv').sample(frac=0.012)
df.head()
```

```python
df.sort_values(by=['species'])
```

```python
df.sort_values(by=['species', 'bill_length_mm'])
```

```python
df.sort_values(
    by=['bill_length_mm', 'bill_depth_mm'],
    key=lambda x: df['bill_depth_mm'] + df['bill_length_mm']
)
```
