# Ejercicio 1

Vamos a crear una clase llamada Persona. Sus atributos son: nombre, edad y DNI.
Construye los siguientes métodos para la clase:
- Un constructor, donde los datos pueden estar vacíos.
- Los setters y getters para cada uno de los atributos. Hay que validar las entradas
de datos.
- mostrar(): Muestra los datos de la persona.
- esMayorDeEdad(): Devuelve un valor lógico indicando si es mayor de edad.

```python
class Persona:
    def __init__(self, nombre='', edad=None, dni=''):
        self.set_nombre(nombre)
        self.set_edad(edad)
        self.set_dni(dni)

    def set_nombre(self, nombre):
        if isinstance(nombre, str) and nombre.strip():
            self._nombre = nombre.strip()
        else:
            raise ValueError("Debes introducir una cadena de texto")
    
    def set_edad(self, edad):
        if isinstance(edad, int) and 0 <= edad <= 120:
            self._edad = edad
        else: 
            raise ValueError("Edad no válida")

    def set_dni(self, dni):
        if isinstance(dni, str) and len(dni.strip()) in (8, 9):
            self._dni = dni.strip()
        else:
            raise ValueError("Deben ser 8 o 9 caracteres")
    
    def get_nombre(self):
        return self._nombre

    def get_edad(self):
        return self._edad

    def get_dni(self):
        return self._dni

    def mostrar(self):
        return f"Nombre: {self.get_nombre()} \nEdad: {self.get_edad()} \nDNI: {self.get_dni()}"

    def es_mayor_de_edad(self):
        return self.get_edad() is not None and self.get_edad() >= 18

persona1 = Persona("Cristóbal", 37, "12345678f")
print(persona1.mostrar())
print(persona1.es_mayor_de_edad())
```

Crea un diccionario de coches donde las claves sean los objetos Coche y los valores sean el dueño del coche. Implementa los métodos __eq__ y __hash__ para hacer esto posible.

```python
class Coche:
    def __init__(self, marca, modelo, año):
        self.marca = marca
        self.modelo = modelo
        self.año = año
    
    def __eq__(self, otro):
        return (
            isinstance(otro, Coche)
            and self.marca == otro.marca
            and self.modelo == otro.modelo
            and self.año == otro.año
        )
    
    def __hash__(self):
        return hash((self.marca, self.modelo, self.año))

coche1 = Coche("Opel", "Corsa", 2000)
coche2 = Coche("Opel", "Corsa", 2000)

coches = {coche1: "Manolo", coche2: "Felipe"}
print(coches[coche1])
```

```python
class Matematica:
    @staticmethod
    def suma(a, b):
        return a + b

    @staticmethod
    def resta(a, b):
        return a - b

    @staticmethod
    def multiplicacion(a, b):
        return a * b

    @staticmethod
    def division(a, b):
        if b != 0:
            return a / b
        else:
            return "Error: división por cero"

print(Matematica.suma(3, 4))         
print(Matematica.division(10, 2))     
print(Matematica.division(8, 0)) 
```

Implementa métodos para añadir y eliminar habitaciones dentro de una casa.
Asegúrate de que si se destruye una Casa, todas las Habitaciones asociadas también se eliminen.

```python
class Habitacion:
    def __init__(self, nombre):
        self.nombre = nombre

class Casa:
    def __init__(self):
        self.habitaciones = []  
    
    def agregar_habitacion(self, nombre):
        habitacion = Habitacion(nombre)
        self.habitaciones.append(habitacion)
    
    def eliminar_habitacion(self, nombre):
        self.habitaciones = [h for h in self.habitaciones if h.nombre != nombre]
```

# Ejercicio 16: Polimorfismo
Descripción: Práctica de polimorfismo con herencia.
Ejercicio: Crea una clase base llamada Figura con un método area(). Luego, crea subclases como Cuadrado, Circulo y Triangulo que sobrescriban el método area() para calcular el área de cada figura.
Implementa un array de objetos de tipo Figura que contenga varias figuras y usa polimorfismo para calcular el área de cada una sin saber el tipo exacto de figura.

```python
import math

class Figura:
    def area(self):
        pass

class Cuadrado(Figura):
    def __init__(self, lado):
        self.lado = lado

    def area(self):
        return self.lado ** 2

class Circulo(Figura):
    def __init__(self, radio):
        self.radio = radio

    def area(self):
        return math.pi * self.radio ** 2

class Triangulo(Figura):
    def __init__(self, base, altura):
        self.base = base
        self.altura = altura

    def area(self):
        return (self.base * self.altura) / 2

cuadrado = Cuadrado(8)
circulo = Circulo(10)
triangulo = Triangulo (8,10)

print(cuadrado.area())
print(circulo.area())
print(triangulo.area())
```
